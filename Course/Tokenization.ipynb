```python
from transformers import BertTokenizer
```

```python
tokenizer = BertTokenizer.from_pretrained("bert-base-uncased")
tokens = tokenizer.tokenize("Hola cómo estás?")
```

```python
tokens
```

```python
import torch
from transformers import BertModel

model = BertModel.from_pretrained("bert-base-uncased")
```

```python
king_token_id = tokenizer.convert_tokens_to_ids(["king"])[0]
king_embedding = model.embeddings.word_embeddings(torch.tensor([king_token_id]))

man_token_id = tokenizer.convert_tokens_to_ids(["man"])[0]
man_embedding = model.embeddings.word_embeddings(torch.tensor([king_token_id]))

woman_token_id = tokenizer.convert_tokens_to_ids(["woman"])[0]
woman_embedding = model.embeddings.word_embeddings(torch.tensor([king_token_id]))

queen_token_id = tokenizer.convert_tokens_to_ids(["queen"])[0]
queen_embedding = model.embeddings.word_embeddings(torch.tensor([queen_token_id]))

dog_token_id = tokenizer.convert_tokens_to_ids(["dog"])[0]
dog_embedding = model.embeddings.word_embeddings(torch.tensor([dog_token_id]))

cat_token_id = tokenizer.convert_tokens_to_ids(["cat"])[0]
cat_embedding = model.embeddings.word_embeddings(torch.tensor([cat_token_id]))

animal_token_id = tokenizer.convert_tokens_to_ids(["animal"])[0]
animal_embedding = model.embeddings.word_embeddings(torch.tensor([animal_token_id]))  
```

```python
dog_embedding
```

```python
cos = torch.nn.CosineSimilarity()
```

```python
similarity = cos(king_embedding, queen_embedding)
print(similarity[0])
```

```python
similarity = cos(king_embedding-man_embedding+woman_embedding, queen_embedding)
print(similarity[0])
```

```python
similarity = cos(man_embedding, dog_embedding)
print(similarity[0])
```

```python
similarity = cos(dog_embedding, cat_embedding)
print(similarity[0])
```
